```python
plt.figure(figsize=(15,50), facecolor='white')
plotnumber =1

for column in data.drop(['Result'],axis=1):
    ax = plt.subplot(12,3,plotnumber)
    sns.countplot(data=data, x=column)
    plt.xlabel(column,fontsize=10)
    plotnumber+=1
plt.show()
```
